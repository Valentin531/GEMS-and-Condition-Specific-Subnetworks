```python
import glob
import os
from gemsembler import lp_example, GatheredModels, read_supermodel_from_json, get_models_with_all_confidence_levels, get_model_of_interest
import json
```

```python
base_dir = "/scratch/rebernig/VR001_GEMsambler/Output/gemsembler_supermodel_without_AGORA2_METT/gram_positive"
output_base_dir = "/scratch/rebernig/VR001_GEMsambler/Output/gemsembler_output/output_gemsembler_supermodel_without_AGORA2_METT/gram_positive"

model_paths = glob.glob(os.path.join(base_dir, "**", "*.json"), recursive=True)

for path in model_paths:
    try:
        bacterial_name = os.path.splitext(os.path.basename(path))[0]
        outdir = os.path.join(output_base_dir, bacterial_name)
        os.makedirs(outdir, exist_ok=True)

        # Use the gemsembler loader!
        supermodel_lp = read_supermodel_from_json(path)

        # Run your analysis
        supermodel_lp.get_all_confidence_levels()
        lp_gems_conflev = get_models_with_all_confidence_levels(supermodel_lp, outdir)


        print(f"Processed and saved submodels for {bacterial_name} to {outdir}")

    except Exception as e:
        print(f"Failed to process {path}: {e}")

print("Done.")
```
