This codes generates a media list, which is similar to the mGAM used by Juan

```python
lb_minus_o2 = [
    "adn","ala__L","amp","arg__L","asp__L","ca2","cbl1","cl","cmp","cobalt2",
    "cu2","cys__L","dad_2","dcyt","fe2","fe3","fol","glc__D","glu__L","gly",
    "gmp","gsn","h2o","h","his__L","hxan","ile__L","ins","k","leu__L",
    "lipoate","lys__L","met__L","mg2","mn2","mobd","na1","nac","nh4","ni2",
    "phe__L","pheme","pi","pnto__R","pro__L","pydx","ribflv","ser__L","so4",
    "thm","thr__L","thymd","trp__L","tyr__L","ump","ura","uri","val__L","zn2"
]

AA = {"ala__L","arg__L","asp__L","cys__L","glu__L","gly","his__L","ile__L",
      "leu__L","lys__L","met__L","phe__L","pro__L","ser__L","thr__L",
      "trp__L","tyr__L","val__L"}
BASE_NUC = {"adn","amp","cmp","dad_2","dcyt","gmp","gsn","hxan","ins","thymd",
            "ump","ura","uri"}
VIT_COFACT = {"thm","ribflv","pydx","fol","pnto__R","cbl1","lipoate","pheme"}
INORG_NONLIM = {"na1","k","cl","ca2","mg2","pi","so4","nh4","h2o","h"}
TRACE = {"fe2","fe3","zn2","mn2","cobalt2","ni2","mobd"}

def default_cap(met):
    if met == "glc__D":
        return 10.0
    if met in AA:
        return 10.0
    if met in BASE_NUC:
        return 5.0
    if met in VIT_COFACT:
        return 1.0
    if met in INORG_NONLIM:
        return 1000.0
    if met in TRACE:
        return 1.0
    return 5.0

# Add "_e" suffix to match extracellular metabolite IDs in the model
vbdt_media = {f"{met}_e": default_cap(met) for met in lb_minus_o2}
```

```python
# Save to CSV for later use
pd.Series(vbdt_media).to_csv("LB_minus_O2_media.csv", header=["uptake"])
```
