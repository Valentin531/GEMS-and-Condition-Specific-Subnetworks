This code replaces each model’s old biomass reaction with a curated one, sets it as the growth objective, and saves the updated model.

```python
import cobra
from cobra.io import read_sbml_model, write_sbml_model
import os
```

```python
# Load the paths
models_dir = "/scratch/rebernig/VR002_Model_Analysis/Output/Curated_Models_Elena/B.Uniformis/Confidence_Levels"
new_biomass_model_path = "/scratch/rebernig/VR001_GEMsambler/Data/final_biomass_as_model.xml"
output_dir = "/scratch/rebernig/VR002_Model_Analysis/Output/Curated_Models_Elena/B.Uniformis/Confidence_Levels_Biomass"

if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    
# Make a list of the submodel names 
model_names = [ "assembly", "carveme", "core2", "core3", "gapseq", "kbase","agora","core4","core5","curated","modelseed"] #agora2 and core4 taken out
```

```python
#load the new biomass model script
biomass_model = read_sbml_model(new_biomass_model_path)
new_biomass_rxn = biomass_model.reactions.get_by_id("Biomass")
```

```python
# Process each model in the list
for model_name in model_names:
    model_file = f"{model_name}.xml"
    model_path = os.path.join(models_dir, model_file)

    model = read_sbml_model(model_path)
    print(f"Processing model: {model_name}")

    # Remove old Biomass reaction if it exists
    if "Biomass" in model.reactions:
        old_biomass = model.reactions.get_by_id("Biomass")
        model.remove_reactions([old_biomass])

    # Copy and add the new Biomass reaction
    new_biomass_copy = new_biomass_rxn.copy()
    model.add_reactions([new_biomass_copy])

    # Set the new Biomass reaction as the model objective
    model.objective = new_biomass_copy

    # Save the updated model to the output directory
    updated_model_path = os.path.join(output_dir, f"{model_name}.xml")
    write_sbml_model(model, updated_model_path)
```
