```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
```

```python
# Load your matrix
df = pd.read_csv("/scratch/rebernig/VR002_Model_Analysis/Output/Curated_Models_Elena/B.Uniformis/Biomass_Viability/Precursor_Produciblity/precursor_producibility_matrix_new.csv", index_col=0)
```

```python
# Create a clustered heatmap with default clustering
my_cmap = ListedColormap(['lightblue', 'darkgreen'])

# Create the clustered heatmap
g = sns.clustermap(
    df,
    cmap=my_cmap,
    figsize=(14, 10),
    linewidths=0.5,
    cbar=False  # Remove colorbar
)

# Remove colorbar if still present
if hasattr(g, 'cax') and g.cax is not None:
    g.cax.set_visible(False)


# Optional: set axis labels
g.ax_heatmap.set_xlabel("B.uniformis Models")
g.ax_heatmap.set_ylabel("Metabolite Precursors")

plt.title("Clustered Producibility Matrix (1=green, 0=grey)", pad=100)
#give the path where it should be saved 
plt.savefig("/scratch/rebernig/VR002_Model_Analysis/Output/Curated_Models_Elena/B.Uniformis/Biomass_Viability/Precursor_Produciblity/clustered_heatmap_new.png", dpi=300, bbox_inches='tight')

plt.show()

```
