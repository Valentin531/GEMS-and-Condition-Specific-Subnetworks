```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load your data
biomass = pd.read_csv("/scratch/rebernig/VR002_Model_Analysis/Output/Curated_Models_Elena/P.Vulgatus/Biomass_Viability/biomass_values.csv", header=None)
biomass.columns = ['model', 'biomass']
biomass = biomass.set_index('model')
```

```python
# Sort by biomass value (optional)
biomass = biomass.sort_values('biomass', ascending=False)

# Bar plot
plt.figure(figsize=(12,5))
biomass['biomass'].plot(kind='bar')
plt.ylabel('Biomass production (FBA objective)')
plt.xlabel('Model')
plt.title('Biomass production across models')
plt.tight_layout()
plt.savefig("/scratch/rebernig/VR002_Model_Analysis/Output/Curated_Models_Elena/P.Vulgatus/Biomass_Viability/biomass_comparison_P_Vulgatus.png", dpi=300, bbox_inches='tight')
plt.show()

```
